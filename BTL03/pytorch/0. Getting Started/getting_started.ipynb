# Import PyTorch

```python
import torch
import torchvision
```

# PyTorch and TorchVision have been imported
# Let's check PyTorch, TorchVision, and CUDA versions

```python
print('PyTorch version:', torch.__version__)
print('TorchVision version:', torchvision.__version__)
print('CUDA version:', torch.version.cuda)
```

# Check whether GPU can be used or not

```python
print('CPU device: cpu')
print('CUDA available:', torch.cuda.is_available())
print('GPU count:', torch.cuda.device_count())

for i in range(torch.cuda.device_count()):
    props = torch.cuda.get_device_properties(i)
    print(
        f'GPU {i}: {props.name} | '
        f'compute capability: {props.major}.{props.minor} | '
        f'total memory: {props.total_memory / (1024 ** 2):.0f} MiB ({props.total_memory} bytes)'
    )

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('Selected device:', device)
```
