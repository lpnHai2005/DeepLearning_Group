```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 50

# Sinh ngẫu nhiên tọa độ tâm
x = np.random.rand(N)
y = np.random.rand(N)

# Sinh màu ngẫu nhiên
colors = np.random.rand(N)

# Sinh bán kính và tính diện tích
area = np.pi * (15 * np.random.rand(N)) ** 2

# Kiểm tra dữ liệu
print(f'+ x: len={len(x)}, [{min(x):.2f}, {max(x):.2f}]')
print(f'+ y: len={len(y)}, [{min(y):.2f}, {max(y):.2f}]')
print(f'+ colors: len={len(colors)}, [{min(colors):.2f}, {max(colors):.2f}]')
print(f'+ area: len={len(area)}, [{min(area):.2f}, {max(area):.2f}]')

plt.scatter(x, y, s=area, c=colors, alpha=0.5)
plt.show()
```
